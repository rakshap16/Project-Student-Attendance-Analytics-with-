#### Section 1: Data Inspection

```python
"""Q1. Display the first 10 rows, the shape of the DataFrame, and the data types of each column."""
df = pd.read_excel("aiml_attendance.xlsx")
print("First 10 rows:")
display(df.head(10))
print("Shape (rows, columns):", df.shape)
print("\nColumn data types:")
print(df.dtypes)
```

```python
"""Q2. Check for missing values in each column. Identify which columns have null values (especially Last Present Date with "Never") and report the count. """
print("Missing values in each column:")
print(df.isnull().sum())
print("\n'Never' values in Last Present Date:")
print((df["Last Present Date"].astype(str).str.strip().str.lower() == "never").sum())
```

```python
"""Q3. Rename the columns to snake_case (e.g., Roll No → roll_no, Student Name → student_name, Attendance % → attendance_pct). Display the updated column list."""
import re
def to_snake_case(column):
    column = column.strip().replace("%", "pct")
    column = re.sub(r"[^A-Za-z0-9]+", "_", column)
    return column.strip("_").lower()

df.columns = [to_snake_case(c) for c in df.columns]
print(df.columns.tolist())
```

```python
"""Q4. Convert Attendance % from string format (e.g., "64%") to a numeric float column (0.64 or 64.0). Also convert Last Present Date to datetime, handling "Never" as NaT."""
df["attendance_pct"] = (
    df["attendance_pct"].astype(str)
    .str.replace("%", "", regex=False)
    .str.strip()
    .replace("", np.nan)
    .astype(float)
)
df["last_present_date"] = pd.to_datetime(
    df["last_present_date"].replace("Never", pd.NaT),
    errors="coerce"
)
print(df[["attendance_pct", "last_present_date"]].head())
print("Never/missing dates converted to NaT:", df["last_present_date"].isna().sum())
```

#### Section 2: Filtering & Selection

```python
"""Q5. Filter and display all students whose Status is "Critical". Sort them by Attendance % in ascending order."""
critical_students = df[df["status"] == "Critical"].sort_values(
    "attendance_pct", ascending=True
)
display(critical_students)
```

```python
"""Q6. Extract all students who have 0 present days (fully absent). List their roll numbers, names, and consecutive absences."""
fully_absent = df[df["present"] == 0]
display(fully_absent[["roll_no", "student_name", "consecutive_absences"]])
```

```python
"""Q7. Using .loc and .iloc, display:
(a) All "Safe" students using .loc
(b) The first 15 rows and columns 2-6 using .iloc """
safe_students = df.loc[df["status"] == "Safe"]
display(safe_students)
#(b)
display(df.iloc[:15, 2:7])
```

```python
"""Q8. Find all students whose Consecutive Absences is greater than or equal to 5. Display their name, status, and last present date."""
absent_5_or_more = df[df["consecutive_absences"] >= 5]
display(absent_5_or_more[["student_name", "status", "last_present_date"]])
```

### Section 3: Grouping & Aggregation 

```python
"""Q9. Group the data by Status and compute: 
a) Count of students 
b) Average attendance %  
c) Maximum consecutive absences"""
status_summary = df.groupby("status").agg(
    student_count=("student_name", "count"),
    average_attendance_pct=("attendance_pct", "mean"),
    max_consecutive_absences=("consecutive_absences", "max")
)
display(status_summary)
```

```python
"""Q10. Create a new column Attendance Category:
≥ 90% → "Excellent"
75-89% → "Good"
60-74% → "Average"
< 60% → "Poor"
Then use groupby() to count students in each category."""
df["attendance_category"] = pd.cut(
    df["attendance_pct"],
    bins=[-0.001, 60, 75, 90, 100],
    labels=["Poor", "Average", "Good", "Excellent"],
    right=False,
    include_lowest=True
)
df["attendance_category"] = df["attendance_pct"].apply(
    lambda x: "Excellent" if x >= 90 else
              "Good" if x >= 75 else
              "Average" if x >= 60 else "Poor"
)
print(df.groupby("attendance_category", observed=False).size())
```

```python
"""Q11. Calculate the total number of classes conducted (sum of Total Classes) and the total present/absent counts across all students using .agg()."""
totals = df[["total_classes", "present", "absent"]].agg("sum")
print(totals)
```

```python
"""Q12. Group by Status and use .agg() to find:
a)min, max, and mean of Present
b)sum of Absent"""
present_absent_summary = df.groupby("status").agg(
    present_min=("present", "min"),
    present_max=("present", "max"),
    present_mean=("present", "mean"),
    absent_sum=("absent", "sum")
)
display(present_absent_summary)
```

### Section 4: Feature Engineering & Transformation

```python
"""Q13. Create a new column Risk Score using the formula:
Risk Score = (Absent * 0.4) + (Consecutive Absences * 0.6)
Display the top 10 students with the highest risk score."""
df["risk_score"] = (
    df["absent"] * 0.4 + df["consecutive_absences"] * 0.6
)
top_10_risk = df.nlargest(10, "risk_score")
display(top_10_risk[["roll_no", "student_name", "absent","consecutive_absences", "risk_score"]])
```

```python
"""Q14. Add a column Classes Needed (Numeric) by extracting the numeric value from the Classes Needed (75%) column (e.g., "5 more" → 5, "✓ Met" → 0)."""
import pandas as pd

col = "Classes Needed (75%)"

df["Classes Needed (Numeric)"] = (
    pd.to_numeric(
        df[col].astype(str).str.extract(r"(\d+)")[0],
        errors="coerce"
    )
    .fillna(0)
    .astype(int)
)

display(
    df[[
        "Student Name",
        "Classes Needed (75%)",
        "Classes Needed (Numeric)"
    ]]
)
```

```python
"""Q15. Using apply() with a lambda function, create a column Attendance Flag that returns:
a)"Defaulter" if attendance < 75%
b)"Safe" otherwise """
df["attendance_flag"] = df["attendance_pct"].apply(lambda x: "Defaulter" if x < 75 else "Safe")
display(df[["student_name", "attendance_pct", "attendance_flag"]])
```

```python
"""Q16. Use pd.cut() to bin students into attendance ranges: [0-50, 50-75, 75-90, 90-100] and label them as ["Critical", "Warning", "Good", "Excellent"]."""
df["attendance_range"] = pd.cut(
    df["attendance_pct"],
    bins=[-0.001, 50, 75, 90, 100],
    labels=["Critical", "Warning", "Good", "Excellent"],
    include_lowest=True,
    right=False
)
df["attendance_range"] = pd.cut(
    df["attendance_pct"],
    bins=[-0.001, 50, 75, 90, 100.001],
    labels=["Critical", "Warning", "Good", "Excellent"],
    include_lowest=True,
    right=False
)
display(df[["student_name", "attendance_pct", "attendance_range"]])
```

### Section 5: Advanced Analysis 

```python
"""Q17. Using pivot_table(), create a summary showing the count of students by Status and Attendance Category (created in Q10)."""
pivot = pd.pivot_table(
    df,
    index="status",
    columns="attendance_category",
    values="student_name",
    aggfunc="count",
    fill_value=0,
    observed=False
)
display(pivot)
```

```python
"""Q18. Identify duplicate student names (e.g., "ANANYA", "DISHA") and display all rows where the base name repeats. Use string manipulation to extract the base name."""
df = pd.read_excel("aiml_attendance.xlsx")
df.columns = df.columns.str.strip()

# Extract base name
df["base_name"] = (
    df["Student Name"]
    .astype(str)
    .str.strip()
    .str.upper()
    .str.extract(r"^([A-Z]+)", expand=False)
)
repeated_bases = df["base_name"].duplicated(keep=False)
display(
    df.loc[
        repeated_bases,
        ["Roll No", "Student Name", "base_name"]
    ].sort_values("base_name")
)
```

```python
""""Q19. Merge this dataset with a hypothetical Student_Details DataFrame containing Roll No, Branch, and Semester. Perform an inner join and display the merged result."""
student_details = pd.DataFrame({
    "Roll No": df["Roll No"].drop_duplicates().head(5).tolist(),
    "branch": ["AIML"] * min(5, df["Roll No"].nunique()),
    "semester": [5] * min(5, df["Roll No"].nunique())
})
merged = pd.merge(df, student_details, on="Roll No", how="inner")
display(merged)
```

```python
"""Q20. Export the following to separate CSV files:
(a) All "Critical" students
(b) Summary statistics grouped by Status
(c) Top 10 students by Risk Score"""


import pandas as pd

# Load the Excel file
df = pd.read_excel("aiml_attendance.xlsx")
df.columns = df.columns.str.strip()

# Convert numeric columns to numbers
df["Attendance %"] = pd.to_numeric(
    df["Attendance %"].astype(str).str.replace("%", "", regex=False),
    errors="coerce"
)

df["Present"] = pd.to_numeric(df["Present"], errors="coerce").fillna(0)
df["Absent"] = pd.to_numeric(df["Absent"], errors="coerce").fillna(0)
df["Consecutive Absences"] = pd.to_numeric(
    df["Consecutive Absences"], errors="coerce"
).fillna(0)

# (a) Export all Critical students
df[
    df["Status"].astype(str).str.strip().str.lower() == "critical"
].to_csv("critical_students.csv", index=False)

# (b) Summary statistics grouped by Status
summary = df.groupby("Status").agg(
    student_count=("Student Name", "count"),
    average_attendance=("Attendance %", "mean"),
    total_present=("Present", "sum"),
    total_absent=("Absent", "sum"),
    max_consecutive_absences=("Consecutive Absences", "max")
)

summary.to_csv("Status_summary.csv")

# (c) Calculate Risk Score
df["Risk Score"] = (
    df["Absent"] * 0.4
    + df["Consecutive Absences"] * 0.6
)

# Export top 10 students by Risk Score
df.sort_values(
    "Risk Score", ascending=False
).head(10).to_csv("top_10_risk_students.csv", index=False)

print("All three CSV files exported successfully!")
```
